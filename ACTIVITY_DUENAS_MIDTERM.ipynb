The range() function

```python
for x in range(6):
  print(x)
for x in range(2, 6):
  print(x)
```

Nested loops

```python
adjective= {"red","big","tasty"}
fruits={"apple","banana","cherry"}

for x in adjective:
  for y in fruits:
    print(x,y)
```

Break statement in while loop

```python
i=1
while i < 6:
  print(i)
  if i ==3:
    break
  i += 1
```

The continue statement

```python
i=0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
```

The else statement

```python
i = 1
while i < 6:
  print(i)
  i +=  1
else:
  print("i is no longer less tahn 6")
```
