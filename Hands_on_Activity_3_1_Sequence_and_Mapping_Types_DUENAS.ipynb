```python
numberlist = [5, 4, 2, 1, 3]

print(len(numberlist))
print(numberlist[0])
print(numberlist[1])
print(numberlist[2])
print(numberlist[3])
print(numberlist[4])
print(numberlist[-1])
print(numberlist[-2])
print(numberlist[-3])
print(numberlist[-4])
print(numberlist[-5])
```

```python

itemlist = [1, -2.0, [1, 2, 3], "Word"]

print(len(itemlist))
print(itemlist[0])
print(itemlist[1])
print(itemlist[2])
print(itemlist[3])

print(len(itemlist[2]))
print(itemlist[2][0])
print(itemlist[2][1])
print(itemlist[2][2])

print(itemlist[-1])
print(itemlist[-2])
print(itemlist[-3])
print(itemlist[-4])

print(len(itemlist[-2]))
print(itemlist[-2][0])
print(itemlist[-2][1])
print(itemlist[-2][2])
print(itemlist[-2][-3])
print(itemlist[-2][-1])
print(itemlist[-2][-1])
```

The len() function is a built-in tool in Python that counts how many items are inside an object, like a list or a string. When you use it on a list, it tells you the total number of elements stored there. It is helpful because it gives you a quick way to know the size of your data without counting it yourself.

```python

numberlist = [5, 4, 2, 1, 3]
itemlist = [1, -2.0, [1, 2, 3], "Word"]

longlist = numberlist + itemlist

print(len(longlist))
print(longlist[:])
print(longlist[:9])
print(longlist[0:])
print(longlist[1:])
print(longlist[2:])

print(longlist[2:5])
print(longlist[5:2])
print(longlist[8:])
print(longlist[9:])
print(longlist[-9:])
print(longlist[-8:])
print(longlist[-8:-7])
print(longlist[-1:])

print(longlist[10:20])
print(longlist[-7:5])

print(longlist[::1])
print(longlist[::2])
print(longlist[1:8:2])
print(longlist[9:1:-1])
print(longlist[-1::1])
print(longlist[-1::-1])

```
