The len() function is a built-in tool in Python that counts how many items are inside an object, like a list or a string. When you use it on a list, it tells you the total number of elements stored there. It is helpful because it gives you a quick way to know the size of your data without counting it yourself.

```python

numberlist = [5, 4, 2, 1, 3]
itemlist = [1, -2.0, [1, 2, 3], "Word"]

longlist = numberlist + itemlist

print(len(longlist))
print(longlist[:])
print(longlist[:9])
print(longlist[0:])
print(longlist[1:])
print(longlist[2:])

print(longlist[2:5])
print(longlist[5:2])
print(longlist[8:])
print(longlist[9:])
print(longlist[-9:])
print(longlist[-8:])
print(longlist[-8:-7])
print(longlist[-1:])

print(longlist[10:20])
print(longlist[-7:5])

print(longlist[::1])
print(longlist[::2])
print(longlist[1:8:2])
print(longlist[9:1:-1])
print(longlist[-1::1])
print(longlist[-1::-1])

```

```python
phonebook = [
    {'id': 1, 'name': 'Jarel Egar', 'contact': '09060611233', 'address': 'Marikina'},
    {'id': 2, 'name': 'Nichollo Galvez', 'contact': '09933575002', 'address': 'Quezon City'}
]

search_id = int(input("Enter ID to search: "))
found = False

for person in phonebook:
    if person['id'] == search_id:
        print(f"Name: {person['name']}")
        print(f"Contact: {person['contact']}")
        print(f"Address: {person['address']}")
        found = True
        break

if not found:
    print("Record not found.")
```
