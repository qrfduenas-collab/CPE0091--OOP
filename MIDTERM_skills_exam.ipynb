```python
asterisk = 5

for i in range(1, asterisk + 1):
    print("*" * i)

for i in range(asterisk - 1, 0, -1):
    print("*" * i)
```

```python
def bracketss(X):
    designated_bracketz = {')': '(', ']': '[', '}': '{'}
    string = []

    for char in X:
        if char in "({[":
         string.append(char)
        elif char in ")}]":
          if not string or string.pop() != designated_bracketz[char]:
                return False

    return len(string) == 0

print(bracketss("{{[]}}"))
print(bracketss("([)]"))
```
